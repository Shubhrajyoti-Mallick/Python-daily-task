14. Create class Rectangle with private length and width. Add method to calculate
area and perimeter.

```python
class Rectangle:
    def __init__(self,length,breadth):
        self.__length=length
        self.__breadth=breadth
    def get_length(self):
        return self.__length
    def get_breadth(self):
        return self.__breadth
        
    def area(self):
        area=self.__length*self.__breadth
        print(f"area of rectangle is { area}")
    def perimeter(self):
        perimeter=2*(self.__length+self.__breadth)
        print(f" perimeter of rectangle is  {perimeter}")
rectangle=Rectangle(5,10)
rectangle.area()
rectangle.perimeter()
```

16. Create multi-level inheritance: Person → Employee → Manager. Override a
method at each level.

```python
class Person:
    def __init__(self,name):
        self.name=name
    def display(self):
        print(" i love to play")
class Employee(Person):
    def __init__(self,name,age):
        super().__init__(name)
        self.age=age
    def display(self):
        super().display()
        print("i love to study")
class Manager(Employee):
    def __init__(self,name,age,gender):
        super().__init__(name,age)
        self.gender=gender
    def display(self):
        super().display()
        print(" i love to paint")
manager=Manager("shubhra",22,"Male")
manager.display()
manager.age
    
```

18. Create abstract class Shape with abstract method area(). Implement it in Circle
and Square using super().

```python
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def  area(self):
        pass
class Circle(Shape):
    def area(self,r):
        super().area()
        area=3.14*r*r
        print("area of circle is :",area)
class Square(Shape):
    def area(self,length):
        super().area()
        area=length*length
        print("area of square is",area)

circle=Circle()
circle.area(6)
square=Square()
square.area(3)

```

27. Write a menu-driven banking system using OOP + File Handling. Store account
details in a file. Handle FileNotFoundError, ValueError, and custom exceptions.

```python
class InsufficientBalance(Exception):
    pass


class BankSystem:
    def __init__(self, name, age, balance=0):
        self.name = name
        self.age = age
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
        else:
            raise ValueError("Amount should be greater than 0")

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                return self.balance
            else:
                raise InsufficientBalance(
                    "Balance is less than amount entered"
                )
        else:
            raise ValueError("Amount should not be less than 0")

    def show(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Balance: {self.balance}")


while True:
    print("1. Create account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Show details")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    try:
        if choice == 1:
            name = input("Enter your name: ")
            age = int(input("Enter age: "))
            balance = int(input("Enter balance to store: "))

            account = BankSystem(name, age, balance)

            with open("store.txt", "w") as file:
                file.write(f"{name},{age},{balance}")

            print("Account created successfully")

        elif choice == 2:
            with open("store.txt", "r") as file:
                data = file.read()

            name, age, balance = data.split(",")

            amount = int(input("Enter amount to deposit: "))

            account = BankSystem(
                name,
                int(age),
                float(balance)
            )

            account.deposit(amount)

            with open("store.txt", "w") as file:
                file.write(
                    f"{account.name},{account.age},{account.balance}"
                )

        elif choice == 3:
            with open("store.txt", "r") as file:
                data = file.read()

            name, age, balance = data.split(",")

            amount = int(input("Enter amount to withdraw: "))

            account = BankSystem(
                name,
                int(age),
                float(balance)
            )

            account.withdraw(amount)

            with open("store.txt", "w") as file:
                file.write(
                    f"{account.name},{account.age},{account.balance}"
                )

        elif choice == 4:
            with open("store.txt", "r") as file:
                data = file.read()

            name, age, balance = data.split(",")

            account = BankSystem(
                name,
                int(age),
                float(balance)
            )

            account.show()

        elif choice == 5:
            break

    except FileNotFoundError:
        print("Account not found. Create an account first.")

    except ValueError as e:
        print(e)

    except InsufficientBalance as e:
        print(e)
```
