Challenge 1 – Complete Data Cleaning
Your manager gives you a messy dataset.
Perform all of the following:
• Load the dataset.
• Display dataset information.
• Find missing values.
• Remove duplicate rows.
• Fill missing values:
o Age → median
o Balance → mean
• Rename:
o Target → CustomerStatus
o AnnualIncome → Income
• Display the first 15 rows.***

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\jyoti\Downloads\Task_dataset.csv")#Load the dataset
print(df)
print("data loaded successfully")
df.info()#Display dataset information
print(" information of data loaded")
```

```python
#finding missing values
df.isnull().sum()
```

```python
#• Remove duplicate rows.
df.drop_duplicates()
```

```python
'''• Fill missing values:
o Age → median
o Balance → mean'''
# finding median of age
print("median of age is:")
median=df["Age"].median()
print(median)
# finding mean of Balance
print("mean of Balance is: ")
mean=df["Balance"].mean()
print(mean)
print("after filling missing value of Age  and Balance with its median and mean is :")
df[["Age","Balance"]].fillna(f"{mean},{median}")
```

```python
'''• Rename:
o Target → CustomerStatus
o AnnualIncome → Income'''
df.rename(columns={"Target":"CustomerStatus","AnnualIncome":"Income"},inplace=True)
```
