```python
marks=[]
for i in range (5):
    m=int(input(f"enter marks of sub{i+1}"))
    marks.append(m)
average=sum(marks)/len(marks)
if average >=60:
    if all(m>40 for m in marks):
        print("pass with good performance")
    else:
        print("marks is average but one or more subject are less than 40")
else:
    print("average is less than 60")
 
```
