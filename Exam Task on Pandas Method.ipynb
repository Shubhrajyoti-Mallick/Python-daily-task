Q9. Fix Age
The Age column contains:
• Valid numbers
• Missing values
• Invalid text such as unknown, N/A, etc.
Convert Age into a proper numeric column.
Invalid values should become missing values.
Then fill missing Age values using the median age.
Use: pd.to_numeric(), median(), fillna()

```python
df["Age"]=pd.to_numeric(df["Age"],errors="coerce")
print(df["Age"])
print("finding meadian of Age")
median=df["Age"].median()
print(median)
print("filling missing value of age with  median")
df["Age"]=df["Age"].fillna(median)
print(df["Age"])
```

```python
df["Annual_Income"]=pd.to_numeric(df["Annual_Income"],errors="coerce")
print(df["Annual_Income"])
print("calculating median of Annual_Income")
median=df["Annual_Income"].median()
print(median)
print("  missing values is filled  using  median")
df["Annual_Income"]=df["Annual_Income"].fillna(median)
print(df["Annual_Income"])

```

```python
Part 3 — Numerical Cleaning
Q11. Clean Account Balance
The Account_Balance column contains invalid strings and missing values.
Clean the column and replace missing values with the mean Account Balance.
Use:
• pd.to_numeric()
• mean()
• fillna()
```

Q21. Clean Active
Active is supposed to be a binary column, but it contains:
Yes
No
Y
N

yes

Transform these into a consistent representation:
