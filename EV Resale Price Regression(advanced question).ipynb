**6. Battery Health Outlier Investigation
Analyze battery_health_pct.
Identify:
• values below a reasonable minimum
• values above 100
• extreme observations
Then decide whether to remove, cap, or retain suspicious values.**

```python
## 6. Battery Health Outlier Investigation

print("Analyzing battery_health_pct")

# Display battery health values
print(df["battery_health_pct"])

# Statistical summary
print(df["battery_health_pct"].describe())


# 1. Values below a reasonable minimum
# Battery health should not be below 0%
print("Values below reasonable minimum (0%):")
print((df["battery_health_pct"] < 0).sum())


# 2. Values above 100%
print("Values above 100%:")
print((df["battery_health_pct"] > 100).sum())


# 3. Find extreme observations using IQR

print("Calculating Q1 and Q3")

Q1 = df["battery_health_pct"].quantile(0.25)
Q3 = df["battery_health_pct"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)


print("Calculating IQR")

IQR = Q3 - Q1

print("IQR:", IQR)


# Calculate lower and upper limits

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)


# Find extreme observations / outliers

extreme_values = df[
    (df["battery_health_pct"] < lower_limit) |
    (df["battery_health_pct"] > upper_limit)
]

print("Extreme observations:")
print(extreme_values)


# 4. Decide what to do with suspicious values

# Values below 0% or above 100% are impossible,
# so replace them with NaN.
import numpy as np

df.loc[df["battery_health_pct"] < 0, "battery_health_pct"] = np.nan
df.loc[df["battery_health_pct"] > 100, "battery_health_pct"] = np.nan

print("Invalid battery health values replaced with NaN.")

'''* **Values below 0%** → physically impossible → replace with `NaN` or remove.
* **Values above 100%** → physically impossible → replace with `NaN` or remove.
* **IQR extreme observations** → statistically unusual, but not automatically wrong.
* If an extreme value is **possible and genuine**, retain it.
* If an extreme value is caused by a **data-entry or measurement error**, remove or correct it.
* **Capping** should only be used when there is a justified reason to impose a specific limit.

**Important:** `Q1` is used for the **IQR outlier calculation**. It is **not** the “reasonable minimum.”'''
```
