1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change
shallow[2][0] = 99. Print both lists and explain the result.

```python
import copy
original=[1, 2, [3, 4], 5]
shallow=copy.copy(original)
shallow[2][0] = 99
print(original)
print(shallow)
#**Here  shallow copy copies the original  outer object but nested objects are not copied  instead it create  reference to original . so changing inner object changes another  copied list,but if we change the outer object then it will not change the  copied lis

```

2.. Do the same as above but use deepcopy instead of shallow copy. Show the difference.

```python
import copy
original=[1, 2, [3, 4], 5]
deep_copy=copy.deepcopy(original)
deep_copy[2][0] = 99
print(original)
print(deep_copy)
# here the  deep copy copies the original object but is completly  independent of original object it copies  both inner nested and outed object  but d no  reference is shared so changing original  object does affect the  deeep copied object and vice versa
```
