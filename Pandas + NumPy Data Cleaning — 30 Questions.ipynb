**16.Clean income_usd by removing currency symbols, commas, and invalid text,
then convert it to a numeric column using pd.to_numeric().**

```python
df["income_usd"]=df["income_usd"].astype("str")
df["income_usd"] = df["income_usd"].str.replace("$", "", regex=False)
df["income_usd"] = df["income_usd"].str.replace(",", "", regex=False)

df["income_usd"] = pd.to_numeric(
    df["income_usd"],
    errors="coerce"
)

print(df["income_usd"])
```

**18.Identify unrealistic age values using NumPy/Pandas and replace them with
NaN.**

```python
import numpy as np
Q1=df["Age"].quantile(0.25)
Q3=df["Age"].quantile(0.75)
IQR=Q3-Q1
lower_limit=Q1-(IQR*1.5)
upper_limit=Q3+(IQR*1.5)
df.loc[(df["Age"]<lower_limit)|(df["Age"]>upper_limit),"Age"]=np.nan
print(df["Age"])


```

**24.Fill missing numerical values using the median of each numerical column.**

```python
num_column = df.select_dtypes(["number"]).columns

for col in num_column:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)
    
    print(f"{col}: Missing values filled with median = {median_value}")
    print(f"after filling '{col}':{df[col]}")
```

**25.Fill missing categorical values using the mode of each categorical column.**

```python
print("CATAGORICAL COLUMNS:")
print(df.select_dtypes(["object"]).columns)
print("-"*100)
for cat in df.select_dtypes(["object"]).columns:
    mode_value=df[cat].mode()[0]
    print(f"after filling missing value of catagorical column {cat} with mode {mode_value} :")
    df[cat]=df[cat].fillna(mode_value)
    print(df[cat])
    print("-"*100)
```
